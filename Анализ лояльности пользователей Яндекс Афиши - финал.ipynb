````python
!pip install sqlalchemy 
!pip install psycopg2 
import pandas as pd
from sqlalchemy import create_engine
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
!pip install phik -U
from phik import phik_matrix

db_config = {'user': 'praktikum_student', # имя пользователя
             'pwd': 'Sdf4$2;d-d30pp', # пароль
             'host': 'rc1b-wcoijxj3yxfsf3fs.mdb.yandexcloud.net',
             'port': 6432, # порт подключения
             'db': 'data-analyst-afisha' # название базы данных
             }

connection_string = 'postgresql://{}:{}@{}:{}/{}'.format(
    db_config['user'],
    db_config['pwd'],
    db_config['host'],
    db_config['port'],
    db_config['db'],
)

engine = create_engine(connection_string)
query = '''
WITH filtr AS (
  SELECT
      f.order_id,
      f.user_id,
      f.created_dt_msk,
      f.created_ts_msk,
      f.currency_code,
      f.revenue,
      f.tickets_count,
      f.device_type_canonical,
      f.service_name,
      e.event_id,
      e.event_name_code,
      e.event_type_main,
      r.region_name,
      c.city_name
  FROM afisha.purchases AS f
  LEFT JOIN afisha.events AS e ON f.event_id = e.event_id
  LEFT JOIN afisha.city AS c ON c.city_id = e.city_id
  LEFT JOIN afisha.regions AS r ON c.region_id = r.region_id
  WHERE (f.device_type_canonical = 'mobile' OR f.device_type_canonical = 'desktop')
    AND e.event_type_main != 'фильм'
)

SELECT user_id,
       device_type_canonical,
       order_id,
       created_dt_msk AS order_dt,
       created_ts_msk AS order_ts,
       currency_code,
       revenue,
       tickets_count,
       created_dt_msk::date - LAG(created_dt_msk::date) OVER (PARTITION BY user_id ORDER BY created_dt_msk) AS days_since_prev,
       event_id,
       event_name_code AS event_name,
       event_type_main,
       service_name,
       region_name,
       city_name
FROM filtr
ORDER BY user_id;
''' 

df = pd.read_sql_query(query, con=engine)
````

**Задача 2.1:** Данные о выручке сервиса представлены в российских рублях и казахстанских тенге. Приведите выручку к единой валюте — российскому рублю.

Для этого используйте датасет с информацией о курсе казахстанского тенге по отношению к российскому рублю за 2024 год — `final_tickets_tenge_df.csv`. Его можно загрузить по пути `https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')`

Значения в рублях представлено для 100 тенге.

Результаты преобразования сохраните в новый столбец `revenue_rub`.

````python
tenge = pd.read_csv('https://code.s3.yandex.net/datasets/final_tickets_tenge_df.csv')

tenge['data'] = pd.to_datetime(tenge['data']).dt.date
df['order_dt'] = pd.to_datetime(df['order_dt']).dt.date

tenge_rates = tenge[['data', 'curs', 'nominal']].rename(columns={'data': 'order_dt'})

df = df.merge(tenge_rates, on='order_dt', how='left')

df['revenue_rub'] = df['revenue']
mask_kzt = df['currency_code'] == 'kzt'
df.loc[mask_kzt, 'revenue_rub'] = df.loc[mask_kzt, 'revenue'] * df.loc[mask_kzt, 'curs'] / df.loc[mask_kzt, 'nominal']

df = df.drop(columns=['curs', 'nominal'])

# df.head(10)
````

````python
def show_missing_stats(df):
    missing_stats = pd.DataFrame({
        'Кол-во пропусков': df.isnull().sum(),
        'Доля пропусков': df.isnull().mean()
    })
    missing_stats = missing_stats[missing_stats['Кол-во пропусков'] > 0]
    
    if missing_stats.empty:
        return "Пропусков в данных нет"
    

    return (missing_stats.style.format({'Доля пропусков': '{:.4f}'}).background_gradient(cmap='coolwarm'))
show_missing_stats(df)
````

````python
overall_share_two = user_profile['is_two'].mean()
print(f'Средняя доля пользователей с 2+ заказами по всей выборке: {overall_share_two:.2%}')
print()


def segment_return_rate(df, group_col, top_n=None):
    grouped = df.groupby(group_col).agg(
        users_count=('user_id', 'count'),
        share_two_plus=('is_two', 'mean')
    ).sort_values('users_count', ascending=False)
    
    if top_n is not None:
        grouped = grouped.head(top_n)
    
    return grouped


def plot_return_rate(grouped, title, xlabel):
    grouped_sorted = grouped.sort_values('share_two_plus', ascending=False)
    
    plt.figure(figsize=(10, 6))
    plt.bar(range(len(grouped_sorted)), grouped_sorted['share_two_plus'], color='skyblue')
    plt.axhline(overall_share_two, color='red', linestyle='--', label='Среднее по выборке')
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel('Доля пользователей с 2+ заказами')
    plt.grid(True, axis='y', alpha=0.5)
    plt.legend()
    
    if len(grouped_sorted) > 10:
        plt.xticks([])
    else:
        plt.xticks(range(len(grouped_sorted)), grouped_sorted.index, rotation=45, ha='right')
    
    plt.show()

event_type_grouped = segment_return_rate(user_profile, 'first_event_type')
print('Доля возврата по типу первого мероприятия:')
print(event_type_grouped)
plot_return_rate(event_type_grouped,
                  'Доля пользователей с 2+ заказами по типу первого мероприятия',
                  'Тип мероприятия')

device_grouped = segment_return_rate(user_profile, 'first_device')
print('Доля возврата по устройству первой покупки:')
print(device_grouped)
plot_return_rate(device_grouped,
                  'Доля пользователей с 2+ заказами по устройству первой покупки',
                  'Устройство')

region_grouped = segment_return_rate(user_profile, 'first_region', top_n=10)
print('Доля возврата по региону первого заказа (топ-10 по размеру сегмента):')
print(region_grouped)
plot_return_rate(region_grouped,
                  'Доля пользователей с 2+ заказами по региону (топ-10)',
                  'Регион')

service_grouped = segment_return_rate(user_profile, 'first_service', top_n=10)
print('Доля возврата по билетному оператору первого заказа (топ-10 по размеру сегмента):')
print(service_grouped)
plot_return_rate(service_grouped,
                  'Доля пользователей с 2+ заказами по билетному оператору (топ-10)',
                  'Билетный оператор')
````

Отсюда можно сделать следующий вывод: если исключить людей, кто покупал всего раз билеты (а таких из предыдущих пунктов: 8363), то пики (как видно из графиков) сохраняются. Значит люди при повторном совершении заказа предпочитают покупать такое же количество билетов, как и в прошлы раз. Это можно объяснить тем, что люди обычно ходят с семьями или близкими друзьями на мероприятия, поэтому количество покупаемых билетов от раза к разу у таких пользователей будет меняться с меньшей вероятностью. Это стоит учитывать. Промежуточные значения понятно откуда появляются: кто-то один в какое-то из мероприятий не может. Поэтому видим, что между целочисленными значениями наблюдаются самые высокие пики в районе среднего (1.5, 2.5, 3.5, 4.5) между двумя ближайшими целыми значениями. Из учёта средних значений и медиан для всех трех вариаций распредления (2.75) чаще всего покупают 3 билета. Большая часть значений сконцентрирована от 1 до 4х билетов, дальше значения сильно уменьшаются - большую компанию людей сложнее собрать

Гипотеза 2. В регионах, где больше всего пользователей посещают мероприятия, выше доля повторных заказов, чем в менее активных регионах. - гипотеза полностью подтвердилась.

Интересное наблюдение, что, чем больше человек совершает заказаов, тем нормальнее распредление таких пользователей. Причём само распределние имеет пик в районе 500 рублей (в районе среднего). Значит, чтобы увеличить число пользователей с возвратом (которые вернулись за повторной покупкой), нужно больше мероприятий с ценой в райное 500 рублей.

Из общего распределения по количетву билетов стало известно, что чаще всего покупают в среднем 3 билета. Значит люди, которые покупают 3 билета - чаще других возвращаются за повторной покупкой.

С понедельника по среду доля вернувшихся - выше среднего - и ещё суббота. Но яркой зависимости не наблюдается.
